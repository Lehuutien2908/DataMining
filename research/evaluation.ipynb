# Dự đoán nguy cơ mắc bệnh tim

Notebook này trình bày toàn bộ quy trình xử lý dữ liệu, xây dựng và so sánh các mô hình (MLP, Random Forest, XGBoost, SVM) nhằm dự đoán khả năng mắc bệnh tim dựa trên các chỉ số lâm sàng.


## Quy trình tổng quát
- Khảo sát dữ liệu, kiểm tra thiếu và ngoại lệ
- Làm sạch, chuẩn hóa và chọn đặc trưng
- Chia dữ liệu train/test (stratify theo nhãn)
- Huấn luyện MLP (Keras Sequential) và các mô hình so sánh: Random Forest, XGBoost, SVM
- Đánh giá Accuracy, Precision, Recall, F1, Specificity, AUC-ROC


```python
import joblib
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.metrics import roc_curve, roc_auc_score

# Load ML models
scaler = joblib.load("../backend/app/models/scaler.pkl")
selector = joblib.load("../backend/app/models/selector.pkl")

import xgboost as xgb
xgb_model = xgb.XGBClassifier()
xgb_model.load_model("../backend/app/models/xgb_model.json")

from tensorflow import keras
mlp_model = keras.models.load_model("../backend/app/models/mlp_model.h5")

# Load dataset for evaluation
df = pd.read_csv("heart.csv")
X = df.drop(columns="target")
y = df["target"]

X_scaled  = scaler.transform(X)
X_sel     = selector.transform(X_scaled)

xgb_probs = xgb_model.predict_proba(X_sel)[:,1]
mlp_probs = mlp_model.predict(X_sel).ravel()

print("AUC XGB:", roc_auc_score(y, xgb_probs))
print("AUC MLP:", roc_auc_score(y, mlp_probs))
```
